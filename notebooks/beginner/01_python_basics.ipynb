## 📚 Python Fundamentals - Start Here If You're New!

Before diving into the tasks, let's understand the basic Python concepts you'll use. **If you're already comfortable with Python, feel free to skip to Task 1.**

### What is a Function?

A **function** is like a recipe: it takes ingredients (inputs), follows steps, and produces a result (output).

```python
# The keyword 'def' means "define a function"
def greet(name):  # 'name' is a parameter (placeholder for input)
    message = "Hello, " + name + "!"  # Do some work
    return message  # The 'return' keyword sends the result back

# Now we can use (call) the function
result = greet("Alice")  # "Alice" is the argument (actual input value)
print(result)  # Output: Hello, Alice!
```

**Key terms:**
- `def` = keyword that defines a function
- `greet` = the function's name
- `name` = **parameter** (the placeholder in the function definition)
- `"Alice"` = **argument** (the actual value you pass when calling)
- `return` = sends a value back to whoever called the function

### What is a List?

A **list** is an ordered collection of items. Think of it like a shopping list or a row of boxes.

```python
# Create a list with square brackets []
temperatures = [72.5, 68, 75, None, 70]

# Access items by index (position). Python counts from 0!
first_temp = temperatures[0]   # Gets 72.5 (first item)
second_temp = temperatures[1]  # Gets 68 (second item)

# Get the length (how many items)
count = len(temperatures)  # 5 items total

# Add items to the end
temperatures.append(73)  # Now temperatures has 6 items
```

### What is a Loop?

A **loop** repeats code multiple times. The `for` loop goes through each item in a collection:

```python
numbers = [1, 2, 3, 4, 5]

# Read this as: "for each number in numbers, do this..."
for num in numbers:
    print(num * 2)  # Prints: 2, 4, 6, 8, 10
```

### What are Conditionals (if statements)?

**Conditionals** let you make decisions in code:

```python
temperature = 85

if temperature > 90:
    print("It's hot!")
elif temperature > 70:  # elif = "else if"
    print("It's warm")
else:
    print("It's cool")
```

**Comparison operators:**
- `==` means "equals" (not to be confused with `=` which assigns values)
- `!=` means "not equals"
- `>`, `<`, `>=`, `<=` compare numbers
- `is` checks if two things are the exact same object (use for `None`)

### What is None?

`None` is Python's special value meaning "nothing" or "no value." It's different from 0, empty string "", or False.

```python
result = None  # This means "no value yet"

if result is None:  # Check if something is None with 'is'
    print("No result available")
```

### Converting Types with float()

`float()` converts numbers and strings into decimal numbers:

```python
x = float("68.5")  # Converts the string "68.5" to the number 68.5
y = float(42)      # Converts the integer 42 to the float 42.0
```

## 💡 Worked Example: Filtering a List

Before you try Task 1, let's see a complete example of how to filter a list. This shows you the pattern you'll use!

**Problem:** Remove all negative numbers from a list and keep only the positive ones.

**Solution:**

````python
# WORKED EXAMPLE: Filter out negative numbers from a list

def keep_positive_numbers(numbers):
    """
    Remove negative numbers and keep only positive ones.
    
    Parameters:
        numbers - a list of integers (can include negative, zero, or positive)
    
    Returns:
        A new list with only the positive numbers (zero is excluded)
    """
    # Step 1: Create an empty list to store the results
    positive_only = []  # This is called an "accumulator" - we build it up as we go
    
    # Step 2: Loop through each number in the input list
    for number in numbers:  # 'for each number in the numbers list...'
        
        # Step 3: Check if the number meets our criteria
        if number > 0:  # Only keep positive numbers (greater than zero)
            
            # Step 4: If it passes the test, add it to our result list
            positive_only.append(number)  # .append() adds to the end of the list
            
        # If the number is negative or zero, we do nothing (skip it)
    
    # Step 5: Return the filtered list
    return positive_only


# Try it out!
mixed_numbers = [5, -2, 0, 8, -10, 3, -1, 12]
result = keep_positive_numbers(mixed_numbers)
print(f"Original: {mixed_numbers}")
print(f"Filtered: {result}")
# Output: [5, 8, 3, 12]


# Let's trace through what happens step-by-step for the first few items:
# 
# Loop iteration 1: number = 5
#   - Is 5 > 0? Yes! 
#   - append(5) → positive_only = [5]
#
# Loop iteration 2: number = -2
#   - Is -2 > 0? No.
#   - Skip it → positive_only = [5]
#
# Loop iteration 3: number = 0
#   - Is 0 > 0? No.
#   - Skip it → positive_only = [5]
#
# Loop iteration 4: number = 8
#   - Is 8 > 0? Yes!
#   - append(8) → positive_only = [5, 8]
#
# ... and so on!


# The pattern you just saw (loop + condition + append) is VERY common!
# You'll use this exact pattern in Task 1 - just with different conditions.
````

## 📖 Python Concepts Reference

Need a quick reminder of what a Python keyword or method does? Look here!

### Keywords

**`def`** - Defines a function
```python
def my_function(parameter):
    return parameter * 2
```

**`return`** - Sends a value back from a function
```python
def add(a, b):
    return a + b  # Sends the sum back to the caller
```

**`if`, `elif`, `else`** - Make decisions
```python
if condition:
    # do this if condition is True
elif other_condition:
    # do this if first was False but this is True
else:
    # do this if all conditions were False
```

**`for`** - Loop through items in a collection
```python
for item in collection:
    print(item)  # Runs once for each item
```

**`in`** - Check if something is in a collection
```python
if 5 in [1, 2, 3, 4, 5]:
    print("Found it!")
```

**`continue`** - Skip to the next iteration of a loop
```python
for num in [1, 2, 3]:
    if num == 2:
        continue  # Skip 2, don't print it
    print(num)  # Prints: 1, 3
```

**`is`** - Check if two things are the exact same object (use for `None`)
```python
if value is None:
    print("No value")
```

**`None`** - Python's special value meaning "nothing" or "no value"

### Common List Methods

**`.append(item)`** - Add an item to the end of a list
```python
numbers = [1, 2, 3]
numbers.append(4)  # Now: [1, 2, 3, 4]
```

**`.extend(other_list)`** - Add all items from another list
```python
numbers = [1, 2]
numbers.extend([3, 4])  # Now: [1, 2, 3, 4]
```

**`len(list)`** - Get the number of items in a list
```python
count = len([1, 2, 3])  # count = 3
```

---

### Common String Methods

**`.lower()`** - Convert string to lowercase
```python
text = "HELLO"
text = text.lower()  # Now: "hello"
```

**`.upper()`** - Convert string to uppercase
```python
text = "hello"
text = text.upper()  # Now: "HELLO"
```

**`.split()`** - Split a string into a list of words
```python
words = "hello world".split()  # Returns: ["hello", "world"]
```

**`.replace(old, new)`** - Replace all occurrences of old with new
```python
text = "hello world".replace("world", "Python")  # "hello Python"
```

### Type Conversions

**`float(x)`** - Convert to decimal number
```python
float("42.5")  # Returns: 42.5
float(42)      # Returns: 42.0
```

**`int(x)`** - Convert to integer (whole number)
```python
int(42.9)    # Returns: 42 (cuts off decimal)
int("42")    # Returns: 42
```

**`str(x)`** - Convert to string (text)
```python
str(42)      # Returns: "42"
str(3.14)    # Returns: "3.14"
```

### Common Errors and What They Mean

**`NameError: name 'x' is not defined`**
- You're trying to use a variable that doesn't exist
- Check your spelling!

**`IndentationError: expected an indented block`**
- Python uses indentation (spaces/tabs) to group code
- After `def`, `if`, `for`, etc., you must indent the next line

**`TypeError: can only concatenate str (not "int") to str`**
- You're trying to combine incompatible types
- Fix: convert types first: `"Number: " + str(42)`

**`IndexError: list index out of range`**
- You're trying to access an item that doesn't exist
- Lists start at index 0 and end at `len(list) - 1`

**`SyntaxError: invalid syntax`**
- Python doesn't understand your code
- Check for typos, missing colons `:`, or mismatched parentheses
