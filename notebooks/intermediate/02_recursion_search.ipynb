## 📚 Quick Lesson: Recursion

**Recursion** is when a function calls itself. It's like a mirror reflecting a mirror - each reflection contains a smaller version of the image.

### What IS Recursion?

Instead of using a loop, a recursive function solves a problem by breaking it into **smaller versions of the same problem**.

**Simple Example: Countdown**

```python
# Non-recursive version (using a loop)
def countdown_loop(n):
    while n > 0:
        print(n)
        n -= 1
    print("Blastoff!")

# Recursive version (function calls itself!)
def countdown_recursive(n):
    if n == 0:  # BASE CASE: stop recursing
        print("Blastoff!")
    else:  # RECURSIVE CASE: call yourself with a smaller problem
        print(n)
        countdown_recursive(n - 1)  # Function calls itself!

countdown_recursive(3)
# Prints: 3, 2, 1, Blastoff!
```

### The Two Essential Parts of Recursion

Every recursive function needs:

**1. Base Case** - The stopping condition (otherwise you get infinite recursion!)
```python
if n == 0:  # STOP! Don't recurse anymore
    return "done"
```

**2. Recursive Case** - Call the function with a simpler/smaller input
```python
else:
    return do_something(n - 1)  # Make the problem smaller
```

### How Recursion Works: The Call Stack

When a function calls itself, Python creates a new "layer" of the function. Think of it like stacking plates:

```python
def factorial(n):
    if n == 1:  # Base case
        return 1
    else:
        return n * factorial(n - 1)  # Recursive case

result = factorial(3)
```

**What happens step-by-step:**

```
1. factorial(3) is called
   → "I need 3 × factorial(2), but I don't know factorial(2) yet..."
   → Pauses and calls factorial(2)
   
2. factorial(2) is called
   → "I need 2 × factorial(1), but I don't know factorial(1) yet..."
   → Pauses and calls factorial(1)
   
3. factorial(1) is called
   → "Base case! I know the answer: 1"
   → Returns 1
   
4. factorial(2) resumes
   → "Now I know factorial(1) = 1, so I can return 2 × 1 = 2"
   → Returns 2
   
5. factorial(3) resumes
   → "Now I know factorial(2) = 2, so I can return 3 × 2 = 6"
   → Returns 6

Final answer: 6
```

This "stacking" of function calls is called the **call stack**.

### Worked Example: Sum of a List

**Problem:** Add all numbers in a list.

**Recursive thinking:** 
- "The sum of a list = the first number + the sum of the rest of the list"
- Keep breaking it down until there's nothing left to sum

```python
def sum_list(numbers):
    # Base case: empty list has sum of 0
    if len(numbers) == 0:
        return 0
    
    # Recursive case: first number + sum of everything else
    first = numbers[0]
    rest = numbers[1:]  # Everything except the first number
    return first + sum_list(rest)

print(sum_list([1, 2, 3, 4]))  # Returns: 10
```

**Trace of execution:**
```
sum_list([1, 2, 3, 4])
  = 1 + sum_list([2, 3, 4])
  = 1 + (2 + sum_list([3, 4]))
  = 1 + (2 + (3 + sum_list([4])))
  = 1 + (2 + (3 + (4 + sum_list([]))))
  = 1 + (2 + (3 + (4 + 0)))  ← Base case!
  = 1 + (2 + (3 + 4))
  = 1 + (2 + 7)
  = 1 + 9
  = 10
```

### When to Use Recursion

Recursion is powerful for problems with:

**1. Tree-like or nested structures** (like Task 1!)
```python
# Nested lists: [1, [2, [3, 4], 5], 6]
# You don't know how deep the nesting goes!
# Recursion handles any depth automatically
```

**2. Problems that break into smaller versions of themselves**
```python
# factorial(5) = 5 × factorial(4)
# fibonacci(n) = fibonacci(n-1) + fibonacci(n-2)
# paths(m, n) = paths(m-1, n) + paths(m, n-1)
```

**3. Graph/tree traversal** (like exploring a maze)

### Common Recursion Pitfalls

**Pitfall 1: Forgetting the base case**
```python
def bad_countdown(n):
    print(n)
    bad_countdown(n - 1)  # Never stops! Crashes with "maximum recursion depth exceeded"
```

**Pitfall 2: Not making the problem smaller**
```python
def bad_sum(numbers):
    if len(numbers) == 0:
        return 0
    return numbers[0] + bad_sum(numbers)  # BUG: numbers never gets smaller!
```

**Pitfall 3: Wrong base case**
```python
def factorial(n):
    if n == 0:  # Should be: if n <= 1
        return 1
    return n * factorial(n - 1)

factorial(-1)  # Infinite recursion! Never reaches 0
```

### Recursion vs Loops

Both can solve the same problems, but:

**Use recursion when:**
- The problem is naturally recursive (trees, nested structures)
- The code is much clearer with recursion

**Use loops when:**
- Simple iteration is all you need
- You want to avoid call stack overhead

**Example - both work:**
```python
# Loop version
def sum_loop(numbers):
    total = 0
    for num in numbers:
        total += num
    return total

# Recursive version
def sum_recursive(numbers):
    if not numbers:
        return 0
    return numbers[0] + sum_recursive(numbers[1:])
```

For simple tasks like summing, the loop is clearer. For nested lists, recursion shines!

## 📚 Quick Lesson: Sets

A **set** is like a list, but with two special properties:
1. **No duplicates** - each item appears only once
2. **Super fast lookups** - checking if something is in a set is O(1), like dictionaries!

### Creating Sets

```python
# Create a set with curly braces {} - but with values only (no key:value pairs)
colors = {"red", "blue", "green"}

# Create an empty set (careful: {} makes an empty dict, not a set!)
empty_set = set()  # Use set() for empty sets

# Convert a list to a set (removes duplicates automatically!)
numbers = [1, 2, 2, 3, 3, 3, 4]
unique = set(numbers)  # {1, 2, 3, 4}
```

### Adding and Removing Items

```python
visited = set()

# Add a single item
visited.add(5)  # Now: {5}
visited.add(10)  # Now: {5, 10}
visited.add(5)  # Still {5, 10} - duplicates are ignored!

# Remove an item
visited.remove(5)  # Now: {10}
# visited.remove(999)  # ERROR if item doesn't exist

# Safe removal (doesn't error if item isn't there)
visited.discard(999)  # No error, set unchanged
```

### Checking Membership (FAST!)

```python
visited = {1, 2, 3, 4, 5}

# Check if an item is in the set - O(1) time!
if 3 in visited:
    print("We've been to 3 before!")

if 10 not in visited:
    print("Haven't visited 10 yet")
```

**Why use sets for "visited" tracking?**
- In maze/graph problems, you need to check "have I been here before?" millions of times
- Sets make this check O(1) instead of O(n) like lists
- For a 1000×1000 grid, this can be 1,000,000× faster!

### Set Operations

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

# Union (all items in either set)
a | b  # {1, 2, 3, 4, 5, 6}

# Intersection (items in both sets)
a & b  # {3, 4}

# Difference (items in a but not in b)
a - b  # {1, 2}

# Length
len(a)  # 4
```

### When to Use Sets

Use sets when:
- You need to track unique items (no duplicates)
- You need fast "is this in the collection?" checks
- You're marking things as "visited" in searches (Task 3!)
- You need set operations (union, intersection, etc.)

```python
# Example: Track visited cells in a maze
visited = set()

position = (0, 0)  # Starting position
visited.add(position)  # Mark as visited

# Later, check if we've been somewhere
new_position = (1, 0)
if new_position not in visited:  # O(1) check!
    print("Exploring new territory!")
    visited.add(new_position)
```

## 📚 Quick Lesson: Stacks and Queues

**Stacks** and **queues** are two ways to organize data. The difference is in which item you get when you remove something.

### Stack - LIFO (Last-In, First-Out)

Think of a **stack of plates**: you add plates to the top, and take plates from the top. The last plate you added is the first one you remove.

```python
# In Python, use a list as a stack
stack = []

# PUSH: Add to the top
stack.append(1)  # Stack: [1]
stack.append(2)  # Stack: [1, 2]
stack.append(3)  # Stack: [1, 2, 3]  ← 3 is on top

# POP: Remove from the top
top = stack.pop()  # Returns 3, stack is now [1, 2]
top = stack.pop()  # Returns 2, stack is now [1]
```

**Visual:**
```
Operation       Stack
---------       -----
push(1)         [1]
push(2)         [1, 2]
push(3)         [1, 2, 3]  ← top
pop()           [1, 2]     (removed 3)
pop()           [1]        (removed 2)
```

**When to use stacks:**
- Depth-First Search (DFS) - explore as deep as possible first
- Undo functionality (undo the most recent action)
- Tracking function calls (call stack!)

### Queue - FIFO (First-In, First-Out)

Think of a **line at a store**: people join at the back, and leave from the front. The first person to arrive is the first to leave.

```python
# In Python, use collections.deque for efficient queues
from collections import deque

queue = deque()

# ENQUEUE: Add to the back
queue.append(1)  # Queue: [1]
queue.append(2)  # Queue: [1, 2]
queue.append(3)  # Queue: [1, 2, 3]  ← back

# DEQUEUE: Remove from the front
front = queue.popleft()  # Returns 1, queue is now [2, 3]
front = queue.popleft()  # Returns 2, queue is now [3]
```

**Visual:**
```
Operation         Queue
---------         -----
enqueue(1)        [1]
enqueue(2)        [1, 2]
enqueue(3)        [1, 2, 3]
                   ↑     ↑
                 front  back
dequeue()         [2, 3]  (removed 1)
dequeue()         [3]     (removed 2)
```

**When to use queues:**
- Breadth-First Search (BFS) - explore layer by layer
- Processing tasks in order (first come, first served)
- Finding shortest paths (BFS guarantees shortest path!)

### Stack vs Queue: A Real Example

**Finding a path in a maze:**

**Using a stack (DFS):**

```python
# Explore in layers - check all neighbors before going deeper
queue = deque([start_position])
visited = set()

while queue:
    position = queue.popleft()  # Get oldest

        return True

        visited.add(position)
        # Add neighbors to explore
        for neighbor in get_neighbors(position):

```

**The difference:**
- **Stack (DFS)**: Plunges deep. Might find a path quickly, but it might not be the shortest.
- **Queue (BFS)**: Explores layer by layer. Finds the shortest path, but explores more positions.

### Quick Reference

| Data Structure | Add Item | Remove Item | Order | Use Case |
|----------------|----------|-------------|-------|----------|
| **List** | `append()` | `pop()` | Any | General storage |
| **Stack** | `append()` | `pop()` | Last-in, first-out | DFS, undo, depth-first |
| **Queue** | `append()` | `popleft()` | First-in, first-out | BFS, tasks, shortest paths |
| **Set** | `add()` | `remove()` | No order (unique) | Fast lookups, tracking |
| **Dict** | `d[key] = val` | `del d[key]` | No order (key→value) | Fast lookups, mappings |
